#FEATURE ENGİNERİNG
DEF short_long stay

```python
def short_long_stay(x):
    if x > 31:
        return "long"
    elif x<=31:
        return "short"
    else:
        return "unkown"
    
```

```python
main_df['shortorlong']=main_df['minimum_nights'].apply(short_long_stay)
```

# BUİLDİNG OUR REGRESSİON MODEL
0 create a predictor matrix(x) and target collumn (y)    Tahmin edici değişkenleri (X) ve hedef değişkeni (y) oluştur

3 handle the null variables (in case by drophing them)     3. Eksik değerleri (NaN) temizle (gerekirse silerek)

```python
#step 0 create x any y object
include_list = ['room_type','neighbourhood_group','price','minimum_nights','number_of_reviews','shortorlong','reviews_per_month']
main_df[include_list]
```

```python
X =main_df[include_list].drop(columns='price')
y= main_df['price']
```

```python
x_dummies=pd.get_dummies(X,dtype="int",drop_first=True)
```

```python
x_dummies.isna().sum()
```

```python
sns.histplot((x_dummies['reviews_per_month']))
```

```python
X_dummies=x_dummies.fillna(0)
X_dummies.isna().sum()
```

#split into training and testing dataset

```python
from sklearn.model_selection  import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X_dummies,y)
```
